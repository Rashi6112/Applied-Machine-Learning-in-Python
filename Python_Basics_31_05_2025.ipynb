# Conditional Statement

keywords:if,else,elif

```python
#if you see a red light go straight else go right
```

```python
#if-when we have one condition to check

#syntax-
#if condition:
     #statement

age=int(input("Enter your age"))
if age>18:
  print("You are eligible to vote")
```

```python
#else:If my other conditions will not be true then my else statement will be executed
age=int(input("Enter your age"))
if age>18:
  print("You are eligible to vote")
else:
  print("You are not eligible to vote")
```

```python
#elif:when we have multiple conditions so we can use elif
age=int(input("Enter your age"))
if age>=13 and age<=19:
  print("You are a teenager")

elif age>20 and age<=50:
  print("You are an adult")

elif age>51:
  print("you are a senior citizen")

else:
  print("You are a minor")
```

```python
#write the code using if else to find whther a number is even or odd.

number=int(input("Enter the number"))
if number%2==0:
  print("The number is even")
else:
  print("The number is odd")
```

# Control Flow statement

Keywords:break,continue,pass

# Break

Stop the execution of code

```python
for i in range(1,11):
  print(i)
  if i==6:
    break
```

# Continue

It skips the current iteration

```python
for i in range(1,11):
  if i==4 or i==7:
    continue
  print(i)
```

# Pass

If you have not decided what you need to do in current iteration then you can use the pass because we cannot keep the scope empty

```python
for i in range(1,11):
  if i==6:
    pass


  print(i)
```
